```python
from ytmusicapi import YTMusic

# Story 1.4: Validation browser.json au démarrage
validate_browser_json()

# Story 1.2: Chargement et validation config
# config = load_config()

# Story 1.3: Authentification YTMusic
ytmusic = YTMusic(auth='browser.json')

print('✓ browser.json valide')
print('✓ config chargée')
print('✓ YTMusic initialisé')

# L'identifiant de la vidéo (trouvable dans l'URL après v=)
video_id = "8zSODUOoE8w" 

try:
    # Pour ajouter à la bibliothèque, on utilise 'LIKE'
    # 'INDIFFERENT' retirerait le titre de la bibliothèque
    response = ytmusic.rate_song(video_id, 'LIKE')
    print(response)
    
    # On vérifie si la réponse contient la confirmation visuelle (le "Toast")
    try:
        # On cherche le texte de confirmation dans la structure complexe
        actions = response.get('actions', [])
        if actions:
            text_response = actions[0]['addToToastAction']['item']['notificationActionRenderer']['responseText']['runs'][0]['text']
            
            if "Saved to liked music" in text_response or "Ajouté aux titres aimés" in text_response:
                print(f"✅ Succès : {text_response}")
            else:
                print(f"Message reçu : {text_response}")
        else:
            print("⚠️ Réponse inattendue, mais l'action a peut-être fonctionné.")
            
    except (KeyError, IndexError):
        # Si la structure est différente, on peut aussi simplement vérifier 
        # si 'actions' existe, ce qui est généralement signe de succès.
        if 'actions' in response:
            print("✅ Succès (confirmé par la présence d'une action dans la réponse).")
        else:
            print("❌ Erreur : La réponse ne contient pas de confirmation.")

except Exception as e:
    print(f"Une erreur est survenue : {e}")
```
